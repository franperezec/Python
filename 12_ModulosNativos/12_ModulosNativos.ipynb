## Uso del Módulo `webbrowser`

El módulo `webbrowser` de Python es una herramienta útil que permite abrir navegadores web directamente desde un script. Esto es particularmente valioso para automatizar tareas repetitivas, como abrir páginas web específicas o acceder a aplicaciones web.

### Ejemplo Básico con `webbrowser`

El siguiente ejemplo muestra cómo utilizar el módulo `webbrowser` para abrir una página web en el navegador predeterminado:

```python
import webbrowser  # Importamos el módulo webbrowser para trabajar con navegadores web

print("Producto encontrado")  # Imprime un mensaje en la consola
webbrowser.open("https://www.google.com")  # Abre el navegador web en la página de Google
```

En este ejemplo, el script imprime un mensaje en la consola y luego abre el navegador web predeterminado en la página de Google. 

### Ejemplo Práctico Ampliado

Vamos a ampliar este ejemplo para que busque un producto en varios sitios web de compras en línea:

```python
import webbrowser  # Importamos el módulo webbrowser para abrir navegadores web

def buscar_producto(producto):
    """
    Función para buscar un producto en diferentes sitios web.
    """
    sitios_web = {
        "Google": f"https://www.google.com/search?q={producto}",
        "Amazon": f"https://www.amazon.com/s?k={producto}",
        "eBay": f"https://www.ebay.com/sch/i.html?_nkw={producto}",
        "MercadoLibre": f"https://listado.mercadolibre.com/{producto}",
    }

    print(f"Buscando '{producto}' en los siguientes sitios web:")

    # Abre el navegador para cada URL en la lista
    for nombre, url in sitios_web.items():
        print(f"Abrir {nombre}: {url}")
        webbrowser.open(url)

# Ejemplo de uso de la función buscar_producto
producto = "teclado mecánico"
buscar_producto(producto)
```

### Explicación del Código

1. **Función `buscar_producto(producto)`**:
   - Esta función toma el nombre de un producto como argumento y crea un diccionario llamado `sitios_web` donde las claves son nombres de sitios web y los valores son URLs construidas dinámicamente usando el nombre del producto.

2. **Abrir Páginas Web**:
   - Itera sobre el diccionario `sitios_web` y utiliza `webbrowser.open(url)` para abrir cada URL en el navegador web predeterminado.

3. **Salida Esperada**:
   - Al ejecutar este código, el navegador web abrirá automáticamente nuevas pestañas para cada sitio web, mostrando los resultados de búsqueda del producto especificado.

## Conclusión

El módulo `webbrowser` de Python es una herramienta versátil para automatizar tareas de navegación web. Ya sea para abrir enlaces específicos, iniciar aplicaciones web, o realizar búsquedas automáticas, `webbrowser` puede hacer que tus scripts sean más interactivos y útiles. En este ejemplo práctico, hemos visto cómo usar `webbrowser` para abrir múltiples sitios web, simplificando tareas repetitivas y mejorando la eficiencia del desarrollo.

## Introducción al Manejo de Fechas en Python

En Python, trabajar con fechas y horas es una tarea común y se facilita gracias a los módulos nativos como `datetime` y `time`. Estos módulos proporcionan herramientas poderosas para obtener la fecha y hora actuales, crear fechas específicas, formatear fechas, comparar fechas y más. A continuación, exploraremos cómo utilizar estos módulos para realizar operaciones básicas de manejo de fechas.

### Módulos Principales para Manejar Fechas

1. **`datetime`**: Este módulo proporciona clases para manipular fechas y horas de manera compleja. Es fundamental para aplicaciones que requieren trabajar con fechas específicas, calcular diferencias de tiempo, o formatear fechas en representaciones legibles.
   
2. **`time`**: Este módulo es más básico y permite trabajar con tiempos en formato de época (Unix timestamp), que es el número de segundos que han transcurrido desde el 1 de enero de 1970. Es útil para medir el tiempo de ejecución de código, o para operaciones que requieren tiempo en un formato numérico simple.

#### Ejemplos Básicos de Uso

##### 1. Obtener el Tiempo Actual en Segundos desde la Época

El tiempo desde la época se refiere al número de segundos que han transcurrido desde el 1 de enero de 1970, conocido como el Tiempo Universal Coordinado (UTC). Esto se puede obtener usando el módulo `time`:

```python
import time  # Importamos el módulo time para trabajar con tiempo en segundos desde la época

# Muestra el tiempo actual en segundos desde la época (1 de enero de 1970) UTC
print(time.time())
```

##### 2. Obtener la Fecha y Hora Actuales

Para obtener la fecha y hora actuales, podemos utilizar el módulo `datetime`:

```python
import datetime  # Importamos el módulo datetime para trabajar con fechas y horas

# Muestra la fecha y hora actual utilizando datetime
print(datetime.datetime.now())
```

O bien, podemos importar directamente la clase `datetime` del módulo `datetime`:

```python
from datetime import datetime  # Importamos la clase datetime del módulo datetime

# Obtenemos la fecha y hora actual
fecha_actual = datetime.now()
print("Fecha y hora actual:", fecha_actual)
```

##### 3. Crear una Fecha Específica

Puedes crear un objeto de fecha específica, como una fecha de cumpleaños, utilizando la clase `datetime`:

```python
# Crear una fecha específica (por ejemplo, fecha de cumpleaños)
fecha_cumple = datetime(2003, 5, 25)  # Creamos una fecha de cumpleaños
```

##### 4. Formatear Fechas

Formatear fechas es útil para presentar fechas de una manera más legible o en un formato específico. Utilizamos el método `strftime()` para formatear fechas:

```python
# Formateamos la fecha actual en un formato específico
fecha_formateada = fecha_actual.strftime("%A, %d de %B de %Y")
print("Fecha formateada:", fecha_formateada)
```

- `%A`: Nombre completo del día de la semana.
- `%d`: Día del mes con dos dígitos.
- `%B`: Nombre completo del mes.
- `%Y`: Año con cuatro dígitos.

##### 5. Convertir una Cadena a un Objeto `datetime`

A veces, las fechas se almacenan como cadenas y necesitamos convertirlas a objetos `datetime` para realizar operaciones. Usamos el método `strptime()` para convertir una cadena a un objeto `datetime`:

```python
# Convertir una cadena a un objeto datetime
fechaString = "25/05/2003"
fecha_cumple2 = datetime.strptime(fechaString, "%d/%m/%Y")
print("Fecha de cumpleaños:", fecha_cumple2)
```

##### 6. Comparar Fechas

Python permite comparar objetos `datetime` directamente utilizando operadores relacionales:

```python
fecha1 = datetime(2021, 5, 25)
fecha2 = datetime(2021, 5, 30)

if fecha1 < fecha2:
    print("fecha1 es anterior a fecha2")
else:
    print("fecha1 es posterior a fecha2")

print(fecha1 < fecha2)  # Devuelve True si fecha1 es anterior a fecha2
```

##### 7. Obtener Componentes Individuales de una Fecha

Podemos obtener componentes individuales como año, mes, día, hora, y minuto de un objeto `datetime`:

```python
print(
    fecha1.year,   # Año
    fecha1.month,  # Mes
    fecha1.day,    # Día
    fecha1.hour,   # Hora
    fecha1.minute  # Minuto
)
```

#### Conclusión

Python proporciona herramientas poderosas y fáciles de usar para trabajar con fechas y horas. Ya sea que necesites obtener la fecha y hora actuales, manipular fechas específicas, convertir cadenas de texto a objetos de fecha, o comparar fechas, los módulos `datetime` y `time` te ofrecen todas las funcionalidades necesarias. Entender cómo manejar fechas y horas en Python es fundamental para cualquier aplicación que requiera cálculos o manipulaciones de tiempo.

## Uso de `timedelta` para Manipular Fechas y Horas en Python

En Python, el módulo `datetime` ofrece varias herramientas para trabajar con fechas y horas. Una de las clases más útiles dentro de este módulo es `timedelta`. Esta clase permite realizar operaciones aritméticas con fechas, como sumar o restar días, horas, minutos, o incluso microsegundos, y calcular diferencias entre fechas. A continuación, exploraremos cómo utilizar `timedelta` para manipular fechas y horas con algunos ejemplos prácticos.

### 1. Creación y Comparación de Fechas

Primero, veamos cómo crear y comparar fechas en Python. Al crear dos objetos de tipo `datetime`, podemos restarlos para obtener un objeto `timedelta` que representa la diferencia entre estas dos fechas.

```python
from datetime import datetime, timedelta  # Importamos las clases necesarias del módulo datetime

# Creamos dos fechas para comparar
fecha1 = datetime(2023, 5, 25)  # 25 de mayo de 2023
fecha2 = datetime(2021, 5, 30)  # 30 de mayo de 2021

# Calculamos la diferencia entre las fechas
diferencia = fecha1 - fecha2
print("Diferencia entre fechas:", diferencia)
print("Días de diferencia:", diferencia.days)  # Días completos entre las fechas
print("Segundos de diferencia:", diferencia.seconds)  # Segundos adicionales que no completan un día
print("Microsegundos de diferencia:", diferencia.microseconds)  # Microsegundos adicionales
print("Total de segundos de diferencia:", diferencia.total_seconds())  # Diferencia total en segundos
print()
```

#### Explicación

- **`diferencia.days`**: Devuelve el número total de días de diferencia entre las dos fechas.
- **`diferencia.seconds`**: Devuelve el número de segundos adicionales (excluyendo los días completos).
- **`diferencia.microseconds`**: Devuelve la fracción restante de la diferencia en microsegundos.
- **`diferencia.total_seconds()`**: Calcula la diferencia total en segundos, incluyendo días completos y cualquier fracción de día.

### 2. Uso de `timedelta` para Añadir Tiempo a una Fecha

`timedelta` también se puede utilizar para sumar o restar tiempo a un objeto `datetime`.

```python
# Sumar 7 días a una fecha específica
fecha3 = datetime(2021, 5, 30) + timedelta(days=7)
print("Fecha original: 2021-05-30")
print("Fecha después de sumar 7 días:", fecha3)
print()
```

#### Explicación

- En este ejemplo, `timedelta(days=7)` crea un período de 7 días, que se suma a `fecha3`, resultando en una nueva fecha.

### 3. Trabajando con la Fecha y Hora Actual

Podemos obtener la fecha y hora actuales y realizar operaciones aritméticas con ellas.

```python
# Obtener la fecha y hora actual
fecha_actual = datetime.now()
print("Fecha y hora actual:", fecha_actual)

# Sumar 7 días a la fecha actual
fecha_futura = fecha_actual + timedelta(days=7)
print("Fecha y hora en una semana:", fecha_futura)

# Calcular la diferencia entre la fecha actual y una fecha específica
fecha_cumple = datetime(2003, 5, 25)  # Fecha de cumpleaños
diferencia = fecha_actual - fecha_cumple
print("Días desde el cumpleaños (25/05/2003):", diferencia.days)
print()
```

### 4. Uso de `timedelta` para Restar Tiempo

También podemos restar tiempo utilizando `timedelta`.

```python
# Restar 2 horas a la fecha y hora actual
fecha_pasada = fecha_actual - timedelta(hours=2)
print("Fecha y hora actual:", fecha_actual)
print("Fecha y hora hace dos horas:", fecha_pasada)
print()
```

#### Explicación

- En este ejemplo, restamos 2 horas de la fecha actual, lo que muestra la hora exacta de hace dos horas.

### 5. Función para Calcular la Edad

Podemos utilizar `timedelta` junto con otros métodos de `datetime` para calcular la edad de una persona en años.

```python
def calcular_edad(fecha_nacimiento):
    """
    Función para calcular la edad de una persona a partir de su fecha de nacimiento.

    Argumentos:
    fecha_nacimiento -- una cadena con la fecha de nacimiento en formato "dd/mm/aaaa"

    Retorna:
    La edad de la persona en años (int)
    """
    # Convertimos la cadena de fecha de nacimiento a un objeto datetime
    fecha_nacimiento = datetime.strptime(fecha_nacimiento, "%d/%m/%Y")
    # Obtenemos la fecha actual
    fecha_actual = datetime.now()
    # Calculamos la diferencia entre la fecha actual y la fecha de nacimiento
    diferencia = fecha_actual - fecha_nacimiento
    # Calculamos la edad en años redondeando hacia abajo
    edad = diferencia.days // 365
    return edad

# Ejemplo de uso de la función calcular_edad
print("Edad de la persona nacida el 25/05/2003:", calcular_edad("25/05/2003"))
```

#### Explicación

- **`strptime()`**: Convierte una cadena de texto a un objeto `datetime`.
- **`diferencia.days // 365`**: Calcula la edad en años dividiendo el número total de días de diferencia por 365.

### Conclusión

Este código demuestra cómo trabajar con fechas y horas en Python utilizando los módulos `datetime` y `timedelta`. Estos ejemplos muestran cómo calcular diferencias de tiempo, sumar o restar días y horas a fechas específicas, y formatear fechas en diferentes formatos. `timedelta` es una herramienta poderosa para cualquier aplicación que necesite manipular o calcular tiempos y fechas. ¡Experimenta cambiando las fechas y los intervalos de tiempo para ver cómo afecta a los resultados!

## Números Aleatorios en Python

Python proporciona el módulo `random` para generar números aleatorios y realizar diversas operaciones relacionadas con la aleatoriedad, como seleccionar elementos aleatorios de una lista, mezclar elementos, y generar contraseñas aleatorias. A continuación, exploraremos cómo usar el módulo `random` para generar números aleatorios y realizar varias operaciones con ejemplos prácticos.

### Generación de Números Aleatorios

1. **Número Decimal Aleatorio entre 0 y 1**:
   
   Usamos `random.random()` para generar un número decimal aleatorio en el rango `[0.0, 1.0)`.

   ```python
   import random  # Importamos el módulo random para generar números aleatorios

   # Generamos un número decimal aleatorio entre 0 y 1
   numero_aleatorio = random.random()
   print("Número aleatorio entre 0 y 1:", numero_aleatorio)
   ```

2. **Número Entero Aleatorio en un Rango Específico**:
   
   Para generar un número entero aleatorio entre dos valores, usamos `random.randint(a, b)`, donde `a` y `b` son los límites del rango (inclusive).

   ```python
   # Generamos un número entero aleatorio entre 0 y 9
   numero_aleatorio_entre = random.randint(0, 9)
   print("Número aleatorio entre 0 y 9:", numero_aleatorio_entre)
   ```

### Selección Aleatoria de Elementos

3. **Elegir un Elemento Aleatorio de una Lista**:

   Utilizamos `random.choice(lista)` para seleccionar un elemento aleatorio de una lista.

   ```python
   elementos = ["rojo", "azul", "verde", "amarillo", "naranja"]  # Lista de elementos para elegir aleatoriamente

   # Elegimos un elemento aleatorio de la lista
   elemento_aleatorio = random.choice(elementos)
   print("Elemento aleatorio de la lista:", elemento_aleatorio)
   ```

4. **Elegir Múltiples Elementos Aleatorios de una Lista (con Repetición)**:

   Con `random.choices(lista, k=n)`, seleccionamos `n` elementos aleatorios de la lista, permitiendo repeticiones.

   ```python
   # Elegir varios elementos aleatorios de la lista
   elementos_aleatorios = random.choices(elementos, k=3)
   print("Varios elementos aleatorios de la lista:", elementos_aleatorios)
   ```

5. **Elegir Múltiples Elementos Aleatorios de una Lista (sin Repetición)**:

   `random.sample(lista, k=n)` selecciona `n` elementos aleatorios de la lista sin permitir repeticiones.

   ```python
   # Elegir varios elementos aleatorios de la lista sin repetición
   elementos_aleatorios_sin_repeticion = random.sample(elementos, k=3)
   print("Varios elementos aleatorios sin repetición:", elementos_aleatorios_sin_repeticion)
   ```

6. **Mezclar Elementos de una Lista**:

   Para mezclar aleatoriamente los elementos de una lista, usamos `random.shuffle(lista)`.

   ```python
   # Mezclamos los elementos de la lista
   random.shuffle(elementos)
   print("Lista de elementos mezclada:", elementos)
   ```

### Generación de Contraseñas Aleatorias

7. **Crear una Contraseña Aleatoria**:

   Utilizando combinaciones de letras, dígitos y símbolos, podemos generar una contraseña aleatoria con `random.choices()`.

   ```python
   import string  # Importamos el módulo string para obtener caracteres alfanuméricos y especiales

   # Longitud de la contraseña
   longitud = 8

   # Caracteres a utilizar para generar la contraseña
   caracteres = string.ascii_letters + string.digits + string.punctuation

   # Generamos la contraseña aleatoria
   password1 = ''.join(random.choices(caracteres, k=longitud))
   print("Contraseña aleatoria:", password1)
   ```

### Uso de Semilla para la Generación de Números Aleatorios

8. **Semilla para la Generación de Números Aleatorios**:

   La función `random.seed()` se utiliza para inicializar el generador de números aleatorios con una semilla específica, lo que permite reproducir resultados aleatorios en diferentes ejecuciones.

   ```python
   # Inicializamos el generador de números aleatorios con una semilla específica
   random.seed(42)

   # Generamos un número entero aleatorio entre 0 y 9
   numero_aleatorio = random.randint(0, 9)
   print("Número aleatorio entre 0 y 9 con semilla 42:", numero_aleatorio)
   ```

### Explicación Adicional

- **Uso de `random.choices()` con cadenas**:
  
  ```python
  # Se crea una lista con 3 elementos aleatorios de la cadena
  print(random.choices("abcdoksdjklfj", k=3))
  ```

  Este ejemplo genera una lista de 3 caracteres aleatorios seleccionados de la cadena proporcionada.

### Conclusión

El módulo `random` en Python es una herramienta poderosa para generar números aleatorios y realizar operaciones relacionadas con la aleatoriedad. Desde la selección de elementos aleatorios hasta la generación de contraseñas seguras, `random` ofrece una amplia variedad de funciones que son útiles en muchas aplicaciones, como simulaciones, juegos, pruebas de software, y más.

## Enviar un Correo Electrónico

En el siguiente ejemplo, mostraremos cómo enviar un correo electrónico que incluye texto y una imagen adjunta utilizando Gmail y Python.

```python
from email.mime.multipart import MIMEMultipart  # Para construir el correo electrónico con múltiples partes
from email.mime.text import MIMEText  # Para incluir texto en el correo
from email.mime.image import MIMEImage  # Para incluir imágenes en el correo
from pathlib import Path  # Para manejar rutas de archivos
import smtplib  # Protocolo de Transferencia Simple de Correo (SMTP)

# Ruta de la imagen que se va a adjuntar
path = Path("12_ModulosNativos/imagen.jpg")

# Crear un objeto MIMEImage con la imagen
mime_image = MIMEImage(path.read_bytes())

# Crear el mensaje de correo electrónico con múltiples partes (texto e imagen)
mensaje = MIMEMultipart()
mensaje["from"] = "Remitente"  # Remitente del correo
mensaje["to"] = "Correo destinatario"  # Destinatario del correo
mensaje["subject"] = "Asunto"  # Asunto del correo

# Crear el cuerpo del mensaje en texto plano
cuerpo_mensaje = MIMEText("Contenido del mensaje", "plain")

# Adjuntar el cuerpo del mensaje al correo
mensaje.attach(cuerpo_mensaje)

# Adjuntar la imagen al correo
mensaje.attach(mime_image)

# Configuración del servidor SMTP de Gmail
with smtplib.SMTP(host="smtp.gmail.com", port=587) as smtp:  # Conectar al servidor SMTP de Gmail usando el puerto 587
    smtp.ehlo()  # Identificarse ante el servidor de correo
    smtp.starttls()  # Iniciar la comunicación encriptada usando TLS (Transport Layer Security)
    smtp.login("tucorreo@gmail.com", "contraseña aplicacion")  # Iniciar sesión en Gmail usando correo y contraseña de aplicación
    smtp.send_message(mensaje)  # Enviar el mensaje de correo electrónico
    print("Mensaje enviado")  # Confirmación de envío
```

## Explicación del Código

1. **Módulos Importados**:
   - `MIMEMultipart`, `MIMEText`, y `MIMEImage`: Utilizados para crear diferentes partes del mensaje de correo (texto, imágenes, etc.).
   - `smtplib`: Proporciona la funcionalidad para enviar correos electrónicos utilizando el protocolo SMTP.
   - `Path`: Facilita la manipulación de rutas de archivos.

2. **Preparación del Mensaje**:
   - Creamos un objeto `MIMEMultipart` que permite agregar diferentes partes al mensaje, como texto e imágenes.
   - Se establece el remitente, el destinatario y el asunto del correo.
   - Se adjunta el cuerpo del mensaje como texto plano (`MIMEText`) y una imagen (`MIMEImage`).

3. **Configuración del Servidor SMTP**:
   - Usamos `smtplib.SMTP` para conectarnos al servidor SMTP de Gmail en el puerto 587.
   - `smtp.ehlo()`: Identificamos nuestra aplicación ante el servidor de correo.
   - `smtp.starttls()`: Habilitamos el cifrado TLS para asegurar la conexión.
   - `smtp.login()`: Nos autenticamos utilizando una dirección de correo electrónico de Gmail y una contraseña de aplicación.
   - `smtp.send_message()`: Enviamos el correo electrónico compuesto.

## Enviar Correos Electrónicos Usando Plantillas HTML en Python

Enviar correos electrónicos con plantillas HTML permite crear mensajes más atractivos y personalizados. En lugar de enviar un simple texto plano, podemos usar HTML para dar formato al correo, añadir imágenes, enlaces y mucho más. Esto es particularmente útil para marketing por correo electrónico o para enviar boletines informativos. En este ejemplo, veremos cómo usar Python para enviar un correo electrónico con una plantilla HTML.

### Herramientas y Módulos Necesarios

1. **`email.mime.multipart` y `email.mime.text`**: Para crear correos electrónicos con múltiples partes (por ejemplo, texto HTML e imágenes).
2. **`email.mime.image`**: Para adjuntar imágenes al correo.
3. **`pathlib.Path`**: Para manejar rutas de archivos.
4. **`string.Template`**: Para manejar plantillas de texto y reemplazar variables dentro de las plantillas.
5. **`smtplib`**: Para enviar correos electrónicos utilizando el protocolo SMTP.

### Crear una Plantilla HTML para el Correo

Podemos utilizar el módulo `string.Template` para crear plantillas HTML dinámicas. Una plantilla HTML puede incluir variables que se reemplazarán con datos específicos cuando se envíe el correo.

```python
from email.mime.multipart import MIMEMultipart  # Para crear correos electrónicos con múltiples partes
from email.mime.text import MIMEText  # Para añadir texto al correo
from email.mime.image import MIMEImage  # Para añadir imágenes al correo
from pathlib import Path  # Para manejar rutas de archivos
from string import Template  # Para manejar plantillas de texto
import smtplib  # Simple Mail Transfer Protocol para enviar correos

# Leer una plantilla HTML desde un archivo
plantilla = Path("12_ModulosNativos/plantilla.html").read_text()

# Crear un objeto Template con la plantilla HTML
template = Template(plantilla)

# Rellenar la plantilla con datos específicos
cuerpo = template.substitute(nombre="Francisco")  # Reemplazamos la variable $nombre por "Francisco"
```

### Construir el Correo Electrónico

Después de crear el cuerpo del correo a partir de la plantilla, construimos el mensaje utilizando `MIMEMultipart`.

```python
# Ruta de la imagen que se va a adjuntar
path = Path("12_ModulosNativos/imagen.jpg")

# Crear un objeto MIMEImage con la imagen
mime_image = MIMEImage(path.read_bytes())

# Crear el mensaje de correo electrónico con múltiples partes
mensaje = MIMEMultipart()
mensaje["from"] = "Remitente"  # Remitente del correo
mensaje["to"] = "CorreoDestino"  # Destinatario del correo
mensaje["subject"] = "Asunto"  # Asunto del correo

# Crear el cuerpo del mensaje en HTML
cuerpo_mensaje = MIMEText(cuerpo, "html")

# Adjuntar el cuerpo del mensaje al correo
mensaje.attach(cuerpo_mensaje)

# Adjuntar la imagen al correo
mensaje.attach(mime_image)
```

### Enviar el Correo Electrónico

Finalmente, utilizamos `smtplib` para conectarnos al servidor SMTP de Gmail y enviar el correo electrónico.

```python
# Configuración del servidor SMTP de Gmail
with smtplib.SMTP(host="smtp.gmail.com", port=587) as smtp:  # Conectar al servidor SMTP de Gmail usando el puerto 587
    smtp.ehlo()  # Identificarse ante el servidor de correo
    smtp.starttls()  # Iniciar la comunicación encriptada usando TLS (Transport Layer Security)
    smtp.login("tucorreo@gmail.com", "contraseña aplicacion")  # Iniciar sesión en Gmail usando correo y contraseña de aplicación
    smtp.send_message(mensaje)  # Enviar el mensaje de correo electrónico
    print("Mensaje enviado")  # Confirmación de envío
```

### Explicación del Código

1. **Cargar la Plantilla HTML**:
   - Utilizamos `Path.read_text()` para leer el contenido de un archivo HTML como texto.
   - `Template(plantilla)` crea un objeto de plantilla a partir del contenido HTML leído.
   - `template.substitute()` reemplaza las variables dentro de la plantilla con los valores proporcionados.

2. **Construcción del Mensaje**:
   - `MIMEMultipart()` se utiliza para crear un contenedor para todas las partes del mensaje, como texto e imágenes.
   - `MIMEText(cuerpo, "html")` se utiliza para crear la parte de texto HTML del correo.
   - `MIMEImage(path.read_bytes())` adjunta una imagen al correo.

3. **Envío del Correo**:
   - Con `smtplib.SMTP()`, nos conectamos al servidor SMTP de Gmail.
   - `smtp.starttls()` habilita el cifrado para la conexión, asegurando que los datos sean transmitidos de manera segura.
   - `smtp.login()` realiza la autenticación en el servidor SMTP utilizando el correo y la contraseña de aplicación.
   - `smtp.send_message(mensaje)` envía el correo electrónico.

### Notas Importantes

- **Seguridad**: Asegúrate de usar contraseñas de aplicación para Gmail para evitar problemas de seguridad.
- **HTML en Correos**: El uso de HTML permite crear correos electrónicos visualmente atractivos con estilos, imágenes, y enlaces.
- **Alternativas**: Considera utilizar servicios de correo como SendGrid, Mailgun o Amazon SES para enviar correos electrónicos masivos o para funciones más avanzadas.
