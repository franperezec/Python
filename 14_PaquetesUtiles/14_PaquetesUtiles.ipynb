## Uso de Variables de Entorno en Proyectos Python

Las variables de entorno son una forma segura y conveniente de gestionar configuraciones que varían entre entornos sin modificar el código fuente. Estas pueden incluir claves secretas, direcciones de servidores, y cualquier otra información que no quieras exponer directamente en tu código.

### Crear un archivo `.env`

1. **Archivo `.env`**: Crea un archivo `.env` en la raíz de tu proyecto para almacenar variables de entorno, como claves API. Por ejemplo:
   ```
   SENDGRID_API_KEY='SG.1234567890'
   ```
   Este archivo no debe ser subido a repositorios públicos para mantener la seguridad de tus claves.

### Leer variables de entorno en Python

2. **Acceder a las variables**:
   En tu script de Python, utiliza el módulo `os` para acceder a las variables de entorno. Crea un archivo, por ejemplo `01_env.py`, y escribe el siguiente código:
   ```python
   import os

   # Obtener el valor de una variable de entorno
   api_key = os.environ.get('SENDGRID_API_KEY')
   print("API Key:", api_key)
   ```

### Configuración y uso de `pipenv`

3. **Instalación de dependencias**:
   Para instalar la biblioteca necesaria, en este caso `sendgrid`, usa:
   ```bash
   pipenv install sendgrid
   ```

4. **Activar el ambiente virtual**:
   Si el comando anterior no reconoce las variables del archivo `.env`, asegúrate de activar el ambiente virtual con:
   ```bash
   pipenv shell
   ```
   Luego, ejecuta tu script para comprobar que la variable de entorno se está leyendo correctamente:
   ```bash
   python 01_env.py
   ```

### Ignorar archivos sensibles

5. **Configurar `.gitignore`**:
   Es crucial configurar un archivo `.gitignore` para evitar que archivos sensibles, como `.env`, sean incluidos en tu control de versiones. Agrega las siguientes líneas a tu `.gitignore`:
   ```
   .env
   ```

### Resumen

Al utilizar variables de entorno, puedes mantener tu código más seguro y fácil de actualizar sin necesidad de cambiar el código fuente. Esta práctica no solo es esencial para la seguridad, sino que también facilita la colaboración y el despliegue de aplicaciones en diferentes entornos.

## Enviar Correos con SendGrid

Enviar correos electrónicos mediante programación puede ser esencial para muchas aplicaciones. Usando SendGrid, un servicio de entrega de correos, puedes integrar fácilmente la funcionalidad de envío de emails en tus aplicaciones Python. A continuación, te explicamos cómo configurar y utilizar SendGrid.

### Configuración de SendGrid

1. **Registro y verificación**:
   - Visita [SendGrid](https://sendgrid.com/en-us) y crea una cuenta.
   - Verifica tu correo electrónico y número de teléfono como parte del proceso de registro.

2. **Autenticación del remitente**:
   - En el dashboard de SendGrid, ve a la sección "Sender Authentication".
   - Selecciona "Verify a Single Sender" y completa los campos requeridos para autenticar tu dirección de correo.

3. **Crear una API Key**:
   - Dirígete a la sección de "API Keys" en el dashboard.
   - Haz clic en "Create API Key", elige "Full Access", asigna un nombre a tu llave y luego clic en "Create & View".
   - Copia la clave generada y guárdala en tu archivo `.env`:
     ```
     SENDGRID_API_KEY="tu_clave_generada"
     EMAIL="tucorreo@gmail.com"
     ```

### Instalación de dependencias

Asegúrate de que `pipenv` esté instalado:
```bash
pip install pipenv
```

### Envío de un mensaje de prueba

4. **Configurar el script de envío**:
   Crea un archivo, por ejemplo `send_email.py`, y escribe el siguiente código:
   ```python
   import os
   from sendgrid import SendGridAPIClient
   from sendgrid.helpers.mail import Mail

   email = os.environ.get('EMAIL')
   mensaje = Mail(
       from_email=email,
       to_emails=email,
       subject='Prueba de envío de correo',
       html_content='<strong>Este es un mensaje de prueba</strong>'
   )

   try:
       api_key = os.environ.get('SENDGRID_API_KEY')
       sg = SendGridAPIClient(api_key)
       response = sg.send(mensaje)
       print("Status Code:", response.status_code)  # Ejemplo: 202
       print("Body:", response.body)
       print("Headers:", response.headers)
   except Exception as e:
       print(e)
   ```

5. **Ejecutar el script**:
   - Si encuentras un error `401 Unauthorized`, es probable que las variables de entorno no se estén leyendo correctamente. Asegúrate de ejecutar tu script dentro de un entorno virtual activado:
     ```bash
     pipenv shell
     python send_email.py
     ```

### Problemas Comunes

- **Error 401 Unauthorized**: Este error suele ocurrir si la API Key no es correcta o no está siendo leída correctamente. Verifica que tu archivo `.env` esté configurado correctamente y que el ambiente virtual donde se ejecuta el script tenga acceso a estas variables.

## Enviar SMS con Twilio

Twilio es una plataforma de comunicaciones que permite enviar mensajes de texto (SMS) entre otros servicios. A continuación, te explicamos cómo configurar y usar Twilio para enviar SMS desde una aplicación Python.

### Configuración Inicial

1. **Registro en Twilio**:
   - Visita [Twilio](https://www.twilio.com/en-us) y crea una cuenta.
   - Sigue el proceso de configuración para obtener tus credenciales: `Account SID` y `Auth Token`.
   - Compra o selecciona un número de teléfono en Twilio que se utilizará para enviar los SMS.

2. **Instalación de la biblioteca Twilio**:
   - Asegúrate de tener `pipenv` instalado y luego ejecuta:
     ```bash
     pip install pipenv  # si aún no está instalado
     pipenv install twilio
     pipenv shell  # para activar el entorno virtual
     ```

### Script para Enviar SMS

3. **Configura el script**:
   - Crea un archivo Python, por ejemplo `send_sms.py`, y escribe el siguiente código:
     ```python
     import os
     from twilio.rest import Client  # Corrige la importación

     # Obtener variables de entorno
     sid = os.environ.get('TWILIO_ACCOUNT_SID')
     token = os.environ.get('TWILIO_AUTH_TOKEN')  # Asegúrate de usar el nombre correcto de la variable
     numero_twilio = os.environ.get('TWILIO_PHONE')  # El número de Twilio

     # Crear cliente de Twilio
     client = Client(sid, token)

     # Enviar el mensaje
     mensaje = client.messages.create(
         to='+15557655310',  # Número del destinatario
         from_=numero_twilio,  # Número de Twilio
         body='Hola desde Twilio'
     )
     print(f'Mensaje enviado: ID {mensaje.sid}')  # Imprime el ID del mensaje para referencia
     ```

### Ejecución del Script

4. **Ejecuta el script**:
   - Asegúrate de estar en el entorno virtual y tener las variables de entorno configuradas correctamente en tu `.env` o de otra forma segura.
   - Ejecuta el script:
     ```bash
     python send_sms.py
     ```

## Uso de la Librería `Requests` en Python para Solicitudes HTTP

`Requests` es una librería de Python que simplifica el envío de solicitudes HTTP/1.1. Es especialmente útil para interactuar con APIs web de manera sencilla.

### Introducción a las APIs REST

Una API REST permite realizar operaciones en servidores utilizando solicitudes HTTP y generalmente devuelve las respuestas en formato JSON, lo que facilita su manipulación en aplicaciones web y móviles.

### Instalación de `Requests`

Primero, instala la librería si aún no lo has hecho:
```bash
pip install requests
```

### Ejemplo Básico: Interacción con la API de GitHub

Aquí tienes un ejemplo que muestra cómo obtener información de la API de GitHub:

```python
import requests

response = requests.get('https://api.github.com')
print(response.status_code)  # Muestra el código de estado HTTP de la respuesta
print(response.json())       # Muestra la respuesta en formato JSON
```

### Códigos de Estado HTTP

Cuando realizas solicitudes utilizando `Requests`, el servidor responde con códigos de estado HTTP. Aquí algunos de los más comunes y su significado:

- **200 OK**: La solicitud ha sido exitosa.
- **201 Created**: Se ha creado un recurso como resultado de una solicitud POST exitosa.
- **204 No Content**: La solicitud ha sido exitosa, pero no hay contenido para devolver.
- **400 Bad Request**: La solicitud no se ha podido entender o procesar.
- **401 Unauthorized**: La autenticación es posible pero ha fallado.
- **403 Forbidden**: La solicitud es válida, pero el servidor se niega a responderla.
- **404 Not Found**: El recurso solicitado no fue encontrado.
- **405 Method Not Allowed**: El método HTTP utilizado no es permitido.
- **500 Internal Server Error**: Error genérico cuando el servidor falla.

### Métodos HTTP con Ejemplos Prácticos

`Requests` soporta varios métodos HTTP para diferentes operaciones:

- **GET**: Solicita datos de un recurso.
- **POST**: Envía datos para crear un nuevo recurso.
- **PUT**: Actualiza un recurso existente.
- **DELETE**: Elimina un recurso.

#### Ejemplo con JSONPlaceholder

Veamos cómo interactuar con JSONPlaceholder, una API para pruebas:

```python
# Obtener usuarios
url = 'https://jsonplaceholder.typicode.com/users'
response = requests.get(url)
print(response.status_code)  # 200
print(response.json())       # Lista de usuarios

# Crear un usuario
user = {"name": "John Doe"}
response = requests.post(url, json=user)
print(response.status_code)  # 201

# Actualizar un usuario
user_update = {"name": "Jane Doe"}
response = requests.put('https://jsonplaceholder.typicode.com/users/1', json=user_update)
print(response.status_code)  # 200

# Eliminar un usuario
response = requests.delete('https://jsonplaceholder.typicode.com/users/1')
print(response.status_code)  # 200
```

### Conclusión

La librería `Requests` es una herramienta poderosa y esencial para los desarrolladores que necesitan interactuar con APIs REST. Proporciona una interfaz simple para enviar todo tipo de solicitudes HTTP y manejar respuestas de manera eficiente.

## Tutorial de Web Scraping con Python y Beautiful Soup

En este tutorial aprenderás a realizar **web scraping** utilizando la librería **BeautifulSoup** para extraer información de páginas web. Para ilustrar el proceso, trabajaremos con un ejemplo real: extraer los títulos de las preguntas y los nombres de usuario que publican en [Stack Overflow](https://stackoverflow.com/questions).

### 1. Introducción

El **web scraping** consiste en **automatizar la extracción de datos** de sitios web. Esta técnica es muy útil para recopilar información de manera más eficiente que hacerlo manualmente. En Python, una de las formas más populares de realizar web scraping es usando las librerías **`requests`** (para realizar solicitudes HTTP) y **`BeautifulSoup`** (para procesar y analizar la respuesta HTML).

En este tutorial:

- Conocerás los pasos fundamentales para realizar scraping.
- Aprenderás a identificar y utilizar selectores CSS para localizar elementos específicos en el HTML.
- Verás cómo iterar y formatear la información extraída.
- Revisarás casos de uso y consejos prácticos para un scraping más robusto.

---

### 2. Instalación de las librerías necesarias

Para realizar web scraping, necesitaremos las siguientes librerías:

- **requests**: permite hacer peticiones HTTP de manera sencilla.
- **beautifulsoup4**: permite analizar y manipular el documento HTML.

En caso de que no las tengas instaladas, puedes hacerlo ejecutando:

```bash
pip install requests beautifulsoup4
```

O bien, si estás usando **pipenv**:

```bash
pipenv install beautifulsoup4
```

> **Nota**: Asegúrate de tener un entorno virtual activo para mantener organizado tu proyecto.

---

### 3. Estructura del código de scraping

A continuación, se muestra el código completo que utilizaremos como referencia. Más adelante, lo analizaremos paso a paso:

```python
import requests
from bs4 import BeautifulSoup

# Definimos la URL a la que haremos la petición
url = 'https://stackoverflow.com/questions'

# Realizamos la petición GET a la URL
respuesta = requests.get(url)

# Obtenemos el texto (HTML) de la respuesta
texto = respuesta.text

# Creamos un objeto BeautifulSoup para poder analizar el HTML
soup = BeautifulSoup(texto, 'html.parser')

# Seleccionamos todas las etiquetas que contengan la clase .s-post-summary
preguntas = soup.select('.s-post-summary')

# Iteramos sobre las preguntas encontradas
for pregunta in preguntas:
    # Extraemos el título de la pregunta
    titulo = pregunta.select_one('.s-link').getText()
    # Extraemos el nombre de usuario que publicó la pregunta
    usuario = pregunta.select_one('.s-user-card--link').getText()
    
    # Imprimimos los resultados
    print(f"Usuario: {usuario.strip()} - Pregunta: \n{titulo.strip()}")
```

---

### 4. Explicación paso a paso

#### 4.1. Realizando la petición y obteniendo la respuesta

```python
url = 'https://stackoverflow.com/questions'
respuesta = requests.get(url)
texto = respuesta.text
```

1. Definimos la URL de la página que queremos extraer, en este caso, la sección de preguntas de Stack Overflow.
2. Usamos `requests.get(url)` para **hacer una solicitud HTTP** y obtener una respuesta.
3. La variable `texto` contendrá el **HTML completo** de la página.

> **Consejo**: Antes de continuar, es útil abrir el HTML en el inspector de tu navegador para familiarizarte con la estructura de la página.

---

#### 4.2. Creando el objeto BeautifulSoup

```python
soup = BeautifulSoup(texto, 'html.parser')
```

- Creamos un objeto **BeautifulSoup** a partir del texto HTML.  
- `html.parser` indica que se usará el *parser* básico de Python. Alternativamente, puedes usar otros parsers como `lxml`, pero debes asegurarte de instalarlos antes.

> **Nota**: BeautifulSoup traduce la estructura HTML en un árbol que podemos navegar para **buscar y extraer** elementos de interés.

---

#### 4.3. Localizando elementos con selectores CSS

```python
preguntas = soup.select('.s-post-summary')
```

- La función `select()` permite usar **selectores CSS** para encontrar elementos en el HTML.
- En este caso, buscamos todos los elementos que tengan la clase `.s-post-summary`.
- El resultado es una **lista** de elementos que coinciden con ese selector.

> **Consejo**: Puedes identificar los selectores en el HTML utilizando las herramientas de desarrollo del navegador (por ejemplo, la pestaña *Elements* en Chrome).

---

#### 4.4. Extrayendo información específica

Dentro del bucle `for`, tomamos cada uno de los objetos `pregunta` y buscamos elementos más específicos:

```python
titulo = pregunta.select_one('.s-link').getText()
usuario = pregunta.select_one('.s-user-card--link').getText()
```

- `select_one(selector)` busca **un solo elemento** que coincida con el selector CSS dado.
- `.getText()` devuelve el **texto interno** de la etiqueta HTML.

**¿Por qué usamos `select_one` en lugar de `select` aquí?**

- Porque esperamos encontrar **solo un** enlace con la clase `.s-link` dentro de cada pregunta. Igualmente, habrá un único usuario con la clase `.s-user-card--link`.

> **Consejo**: Usar `strip()` para limpiar espacios en blanco al principio o al final del texto.

---

#### 4.5. Imprimiendo los resultados

```python
print(f"Usuario: {usuario.strip()} - Pregunta: \n{titulo.strip()}")
```

- Con la información extraída, formateamos un mensaje y lo mostramos en consola.
- La función `strip()` se encarga de eliminar espacios en blanco adicionales.

Al ejecutar el script, obtendrás una lista de preguntas con el nombre de usuario y el título de la pregunta publicada en Stack Overflow.

## Pruebas Automáticas e Instalación de Selenium

En este cuaderno aprenderás los pasos necesarios para **instalar Selenium**, configurar los **drivers** de tu navegador preferido (ejemplo: **Chrome**) y crear las bases de un entorno para **pruebas automáticas** en Python. Al final, tendrás una visión general de cómo iniciar con la automatización de navegadores.

---

### 1. ¿Qué es Selenium?

[Selenium](https://www.selenium.dev/) es un conjunto de herramientas y bibliotecas que permiten automatizar navegadores web. Es muy utilizado para:

- Realizar **pruebas automáticas** (test) de aplicaciones web.
- Automatizar tareas repetitivas, como extraer datos o validar formularios.
- Ejecutar **scripts** de forma controlada en diferentes navegadores (Chrome, Firefox, Safari, etc.).

> **Nota**: Existen varias bibliotecas para usar Selenium en distintos lenguajes (Java, C#, Python, etc.). Aquí nos enfocaremos en **Python**.

---

### 2. Instalación de Selenium

1. Abre tu terminal o consola (Command Prompt en Windows, Terminal en macOS o Linux).
2. Asegúrate de tener **Python** instalado y en tu variable PATH.
3. Instala **Selenium** utilizando `pip`:
   ```bash
   pip install selenium
   ```
4. Puedes verificar la instalación con:
   ```bash
   pip show selenium
   ```
   o importándolo en un intérprete de Python:
   ```python
   python
   >>> import selenium
   >>> selenium.__version__
   ```

---

### 3. ¿Por qué se necesita un **WebDriver**?

Selenium se basa en controladores (drivers) específicos para cada navegador web. Estos controladores permiten que Selenium “controle” el navegador:

- **ChromeDriver** para Google Chrome
- **GeckoDriver** para Mozilla Firefox
- **SafariDriver** para Safari
- **EdgeDriver** para Microsoft Edge

En este cuaderno nos centraremos en **ChromeDriver**, pero la configuración es similar para otros navegadores.

---

### 4. Descarga del ChromeDriver

#### 4.1. Verificar Versión de Chrome

1. Abre Google Chrome.
2. Ve a `Configuración > Información de Chrome`.
3. Identifica el número de versión. Por ejemplo: **131.0.6778.205** (versión de ejemplo).

#### 4.2. Descargar ChromeDriver Compatible

- Visita el sitio oficial: [https://chromedriver.chromium.org/downloads](https://chromedriver.chromium.org/downloads)
- Si tu versión es muy reciente, dirígete a [Chrome for Testing](https://googlechromelabs.github.io/chrome-for-testing/) para descargar una versión compatible.
- Selecciona el **sistema operativo** adecuado y descarga el archivo comprimido (`zip`).

Por ejemplo, para Windows 64 bits con la versión de Chrome *131.0.6778.205*:
```
chromedriver_win64.zip
```
En la página de Chrome for Testing encontrarás el enlace directo, por ejemplo:
```
https://storage.googleapis.com/chrome-for-testing-public/131.0.6778.204/win64/chromedriver-win64.zip
```
(El número de versión puede variar en tu caso.)

#### 4.3. Extraer y Ubicar el Archivo

Descomprime el archivo `chromedriver-win64.zip` (o el que corresponda a tu sistema operativo) y coloca el **ejecutable** `chromedriver` (o `chromedriver.exe` en Windows) en una carpeta. Por ejemplo:

- **Windows**: `C:\webdriver`
- **macOS**: `/usr/local/bin`
- **Linux**: `/usr/local/bin`

> **Consejo**: En Windows, crea la carpeta `C:\webdriver` para mantener el ejecutable del `chromedriver`.

---

### 5. Configuración de Variables de Entorno

Para que el sistema reconozca el `chromedriver` desde cualquier ubicación, debes configurar tu **variable PATH**.

#### 5.1. Windows

1. En el **Buscador** (Inicio), escribe: `Environment Variables` o `Variables de Entorno`.
2. Selecciona **Editar las variables de entorno del sistema**.
3. En la ventana que aparece, haz clic en **Variables de Entorno**.
4. En el panel **Variables de usuario** o **Variables del sistema**, busca la variable `Path` y haz doble clic.
5. Agrega una nueva entrada con la ruta donde se encuentra tu `chromedriver`, por ejemplo:
   ```
   C:\webdriver
   ```
6. Presiona **Aceptar** en todas las ventanas.

Ahora Windows sabrá dónde encontrar `chromedriver.exe`.

#### 5.2. macOS

1. Abre la aplicación **Terminal**.
2. Verifica que el archivo `chromedriver` esté en `/usr/local/bin`. Si no es así, muévelo o cópialo:
   ```bash
   cd /ruta/donde/se/descargo/
   cp chromedriver /usr/local/bin
   ```
3. Asegúrate de que `/usr/local/bin` forme parte de tu PATH. Para ello, puedes editar el archivo de configuración de tu shell (por ejemplo, `~/.bash_profile`, `~/.zshrc`) y agregar:
   ```bash
   export PATH="/usr/local/bin:$PATH"
   ```
4. Abre una nueva ventana de Terminal o ejecuta `source ~/.bash_profile` (o `~/.zshrc`) para recargar la configuración.

> **Nota**: Si el sistema detecta que `chromedriver` es una aplicación descargada de internet, puede solicitar permisos especiales. Selecciona **Open Anyway** en la ventana emergente.

---

### 6. Validando la Instalación

Para comprobar que el **ChromeDriver** está correctamente instalado y reconocido, abre tu consola y teclea:

```bash
chromedriver --version
```

Deberías ver algo como:
```
ChromeDriver 131.0.6778.205 ...
```
Si aparece un **“comando no reconocido”**, revisa que tu ruta esté bien definida en la variable PATH o que hayas copiado el archivo en la ubicación correcta.

### 7. Primer Script con Selenium

Una vez que tengas **Selenium** y **ChromeDriver** configurados, puedes probar un script sencillo en Python:

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

# 1. Iniciar el navegador
driver = webdriver.Chrome()  # ChromeDriver debe estar en tu PATH

# 2. Navegar a una página
driver.get("https://www.google.com")

# 3. Esperar unos segundos para ver la página
time.sleep(3)

# 4. Localizar la barra de búsqueda y escribir algo
search_bar = driver.find_element(By.NAME, "q")
search_bar.send_keys("Hola mundo")
search_bar.submit()

# 5. Esperar para ver los resultados
time.sleep(5)

# 6. Cerrar el navegador
driver.quit()
```

- **webdriver.Chrome()** inicia una instancia de Chrome.
- **driver.get(...)** navega a la URL especificada.
- **driver.find_element(...)** busca un elemento en la página.  
- **send_keys(...)** escribe en el campo de texto.
- **submit()** envía la búsqueda.

> **Consejo**: Recuerda que **time.sleep()** se usa solo como pausa simple para ver lo que ocurre. En pruebas automáticas reales, se recomienda usar **esperas explícitas** de Selenium (por ejemplo, `WebDriverWait`) para sincronizar con exactitud.

## Tutorial: Automatización de Navegador Web con Selenium

En este tutorial aprenderás los fundamentos para **automatizar interacciones con un navegador** usando la librería [Selenium](https://www.selenium.dev/) en Python. Nuestro ejemplo práctico consiste en **iniciar sesión** en GitHub de forma automática y segura, aprovechando el uso de **variables de entorno** para proteger tus credenciales.

---

### 1. Introducción

Selenium es una herramienta popular que permite **controlar navegadores** (Chrome, Firefox, Safari, etc.) a través de scripts de automatización. Es muy útil en:

- **Pruebas automáticas**: Verificar el comportamiento de aplicaciones web.
- **Web scraping avanzado**: Interactuar con sitios que requieren acciones complejas (rellenar formularios, pulsar botones, etc.).
- **Tareas repetitivas**: Iniciar sesión, descargar archivos o rellenar formularios de forma automatizada.

En este tutorial, te mostraremos cómo usar Selenium para abrir Google Chrome y **realizar un login** en GitHub. Además, aprenderás a **proteger tus credenciales** mediante variables de entorno y la librería `python-dotenv`.

---

### 2. Requisitos de Instalación

Antes de empezar, asegúrate de:

1. **No** nombrar tu script como `selenium.py` para evitar conflictos con la propia librería.
2. **Instalar** Selenium en tu entorno de desarrollo:
   ```bash
   pip install selenium
   ```
   o bien, si usas *pipenv*:
   ```bash
   pipenv install selenium
   ```
3. **Instalar** la librería `python-dotenv` para el manejo seguro de variables de entorno:
   ```bash
   pipenv install python-dotenv
   ```
4. **Tener instalado** y **configurado** un WebDriver compatible con tu navegador (por ejemplo, [ChromeDriver](https://chromedriver.chromium.org/downloads) para Google Chrome).

> **Nota**: Si usas *ChromeDriver*, asegúrate de que coincida con la versión instalada de tu navegador.

---

### 3. Estructura y Configuración del Código

A continuación, se muestra la estructura básica del script, organizada en secciones que facilitan su lectura y mantenimiento.

#### 3.1. Importaciones y Carga de Entorno

```python
# Importaciones principales de Selenium
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support import expected_conditions as EC

# Importaciones para manejo de variables de entorno
import os
from dotenv import load_dotenv

# Cargar variables de entorno desde archivo .env
load_dotenv()
```

- **webdriver, By, WebDriverWait, Keys, EC**: Son componentes de Selenium para iniciar el navegador, localizar elementos y configurar esperas.
- **load_dotenv()**: Carga las variables de entorno definidas en un archivo `.env`.
- **os**: Para acceder a las variables de entorno a través de `os.environ.get("NOMBRE_VARIABLE")`.

> **Consejo**: Almacena tus credenciales en un archivo `.env` (no lo subas a repositorios públicos) y así mantendrás tu información confidencial segura.

---

#### 3.2. Configuración del Navegador

```python
options = webdriver.ChromeOptions()
options.add_experimental_option("detach", True)
browser = webdriver.Chrome(options=options)

browser.implicitly_wait(10)  # Espera implícita de 10 segundos
```

1. `webdriver.ChromeOptions()`: Permite configurar opciones adicionales para Chrome.
2. `add_experimental_option("detach", True)`: Hace que el navegador permanezca abierto incluso después de que el script finalice.
3. `webdriver.Chrome(options=options)`: Crea una instancia de Chrome con la configuración anterior.
4. `browser.implicitly_wait(10)`: Indica que Selenium esperará **hasta 10 segundos** para que aparezcan elementos antes de lanzar un error. Es una espera **implícita**.

> **Nota**: Si prefieres cerrar el navegador inmediatamente, elimina o comenta la opción `"detach"`.

---

#### 3.3. Navegación a GitHub

```python
browser.get("https://www.github.com")

link = browser.find_element(By.LINK_TEXT, "Sign in")
link.click()
```

1. `browser.get("URL")`: Indica la dirección web a la que quieres navegar, en este caso, la página principal de GitHub.
2. `find_element(By.LINK_TEXT, "Sign in")`: Localiza el enlace “Sign in” usando el **texto del enlace**.
3. `click()`: Simula un clic sobre ese elemento.

Este paso nos lleva a la página de inicio de sesión de GitHub.

---

#### 3.4. Interacción con el Formulario de Login

```python
user_input = browser.find_element(By.ID, "login_field")
password_input = browser.find_element(By.ID, "password")

user_input.send_keys(os.environ.get("gh_user"))
password_input.send_keys(os.environ.get("gh_password"))
password_input.send_keys(Keys.RETURN)
```

1. Localizamos los **campos de usuario** y **contraseña** mediante sus IDs (`login_field` y `password`).
2. `send_keys(...)`: Envía texto al campo correspondiente.  
3. `os.environ.get("gh_user")` y `os.environ.get("gh_password")`: Toman los valores definidos en el archivo `.env`.
4. `Keys.RETURN`: Simula la tecla *Enter*, para enviar el formulario.

> **Buena práctica**: Evitar almacenar contraseñas en el código fuente. Usa archivos `.env` y descártalos en tu `.gitignore`.

---

#### 3.5. Verificación del Login Exitoso

```python
try:
    wait = WebDriverWait(browser, 10)  # Espera explícita de 10 segundos
    user_element = wait.until(
        EC.presence_of_element_located(
            (By.CSS_SELECTOR, ".user-mention[href$='/adalovelace-ec']"))
    )
    print(f"¡Login exitoso! Usuario conectado: {user_element.text}")
except Exception as e:
    print("No se pudo verificar el usuario")
    print("URL actual:", browser.current_url)
```

- **WebDriverWait**: Configura una espera **explícita** para que Selenium aguarde hasta que un elemento específico aparezca en la página.
- `EC.presence_of_element_located(...)`: Condición que verifica que un elemento esté presente en el DOM.
- `By.CSS_SELECTOR, ".user-mention[href$='/adalovelace-ec']"`: En este ejemplo, se busca un elemento con la clase `.user-mention` cuyo atributo `href` termine en `/adalovelace-ec`.  
  *(Asegúrate de reemplazarlo con tu usuario o algún selector apropiado a tu cuenta.)*

Si pasados 10 segundos no se encuentra ese elemento, se lanza una excepción, que manejamos con el bloque `try-except`.

---

#### 3.6. Cierre o Persistencia del Navegador

Por defecto, podrías finalizar el script y **cerrar** el navegador con:

```python
browser.quit()
```

No obstante, en el ejemplo, se deja el navegador abierto (`detach=True`) para que puedas inspeccionar la sesión iniciada o realizar más acciones de forma manual.

### 4. Código Completo

A continuación, se presenta el código completo con comentarios adicionales:

```python
# =================================================================
# AUTOMATIZACIÓN DE NAVEGADOR WEB CON SELENIUM
# =================================================================
"""
Este script demuestra cómo usar Selenium para automatizar interacciones
con un navegador web, específicamente para realizar un inicio de sesión
en GitHub de forma automática y segura.
"""

# SECCIÓN 1: IMPORTACIONES Y CONFIGURACIÓN INICIAL
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support import expected_conditions as EC

import os
from dotenv import load_dotenv
load_dotenv()  # Cargamos variables de entorno desde .env

# SECCIÓN 2: CONFIGURACIÓN DEL NAVEGADOR
options = webdriver.ChromeOptions()
options.add_experimental_option("detach", True)
browser = webdriver.Chrome(options=options)
browser.implicitly_wait(10)

# SECCIÓN 3: NAVEGACIÓN Y LOGIN EN GITHUB
browser.get("https://www.github.com")
link = browser.find_element(By.LINK_TEXT, "Sign in")
link.click()

# SECCIÓN 4: INTERACCIÓN CON FORMULARIO DE LOGIN
user_input = browser.find_element(By.ID, "login_field")
password_input = browser.find_element(By.ID, "password")
user_input.send_keys(os.environ.get("gh_user"))
password_input.send_keys(os.environ.get("gh_password"))
password_input.send_keys(Keys.RETURN)

# SECCIÓN 5: VERIFICACIÓN DE LOGIN EXITOSO
try:
    wait = WebDriverWait(browser, 10)
    user_element = wait.until(
        EC.presence_of_element_located(
            (By.CSS_SELECTOR, ".user-mention[href$='/adalovelace-ec']"))
    )
    print(f"¡Login exitoso! Usuario conectado: {user_element.text}")
except Exception as e:
    print("No se pudo verificar el usuario")
    print("URL actual:", browser.current_url)

# SECCIÓN 6: LIMPIEZA (opcional)
# browser.quit()
```
