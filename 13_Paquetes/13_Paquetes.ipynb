# Paquetes en Python

## Introducción

En esta sección, exploraremos cómo utilizar y gestionar paquetes en Python, una habilidad esencial para cualquier desarrollador. Los paquetes son herramientas o bibliotecas que otros desarrolladores han creado para facilitar ciertas tareas en Python. Aprenderemos a encontrar, instalar, y publicar estos paquetes utilizando el Índice de Paquetes de Python (PyPI).

## PyPI - The Python Package Index

PyPI ([https://pypi.org/](https://pypi.org/)) es un repositorio centralizado donde puedes encontrar, instalar y publicar paquetes de Python. Al día de hoy, 24 de agosto de 2024, contiene más de 564,461 proyectos. PyPI facilita la búsqueda e instalación de software desarrollado y compartido por la comunidad de Python y es una herramienta invaluable para cualquier programador.

### Ejemplos de paquetes populares
- **Django**: Un framework de alto nivel para desarrollo web en Python.
- **FastAPI**: Framework moderno y rápido para construir APIs con Python 3.7+.
- **Requests**: Una biblioteca elegante y simple para realizar solicitudes HTTP en Python.

Para encontrar Django, por ejemplo, puedes buscar la frase "A high-level Python web framework..." en PyPI.

## Instalar dependencias 

### Uso básico de pip

Para trabajar con paquetes, usaremos `pip`, el instalador de paquetes para Python:

- Para instalar un paquete:
  ```bash
  pip install requests
  ```

- Para verificar la versión de pip:
  ```bash
  pip --version
  ```

- Para listar los paquetes instalados:
  ```bash
  pip list
  ```

### Semántica de versiones
Cuando ves un número de versión como `requests 2.28.2`:
- `2`: Versión mayor, cambios importantes que pueden incluir incompatibilidades.
- `28`: Versión menor, adiciones de funcionalidad en una manera compatible con versiones anteriores.
- `2`: Parche, correcciones de errores y mejoras de rendimiento sin cambios en la funcionalidad.

### Instalación de versiones específicas
En sistemas MacOS, se suele utilizar `pip3`:
- Para desinstalar un paquete:
  ```bash
  pip uninstall requests
  ```
- Para instalar una versión específica:
  ```bash
  pip install requests==2.18.1
  ```

- Para instalar la última versión compatible dentro de una serie mayor:
  ```bash
  pip install 'requests~=2.18.1'
  ```

Nota: Si usas zsh y encuentras problemas con el formato, encierra la versión entre comillas simples.

## Crear un ambiente virtual en Python

El uso de un ambiente virtual en Python es una práctica recomendada para manejar dependencias y asegurar que tu proyecto sea reproducible sin afectar otros proyectos o el sistema operativo en general. Aquí te explicamos cómo configurar uno paso a paso.

### 1. Limpiar la terminal
Antes de comenzar, es una buena idea limpiar tu terminal para evitar confusión:
```bash
clear
```

### 2. Crear el ambiente virtual
Para crear un ambiente virtual, utiliza el módulo `venv` que viene incluido en Python:
```bash
python -m venv env
```
Este comando crea una nueva carpeta llamada `env` en tu directorio actual, que contendrá el ambiente virtual.

### 3. Activar el ambiente virtual
Una vez creado el ambiente, debes activarlo para empezar a usarlo. El comando para activar el ambiente varía dependiendo del sistema operativo:

- **Windows**:
  ```bash
  env\bin\activate.bat
  ```
- **Linux/Mac**:
  ```bash
  source env/bin/activate
  ```

Una vez activado, notarás que el nombre del ambiente virtual (en este caso, `env`) aparece al inicio de tu línea de comando, indicando que cualquier paquete que instales usando `pip` será colocado en este ambiente aislado, sin afectar el sistema global o otros proyectos.

### 4. Estructura del ambiente virtual
Dentro de la carpeta `env`, encontrarás varias subcarpetas importantes:
- **bin/**: Contiene los ejecutables de Python y los scripts, como `activate`.
- **include/**: Almacena los archivos de encabezado C que Python necesita para compilar algunas dependencias.
- **lib/**: Aquí se instalan las dependencias del proyecto.
- **pyvenv.cfg**: Un archivo de configuración para el ambiente virtual.

### 5. Ver el código fuente de las dependencias
Una ventaja de usar ambientes virtuales es que puedes acceder fácilmente al código fuente de las dependencias instaladas, lo cual es útil para el aprendizaje y la depuración.

### 6. Desactivar el ambiente virtual
Cuando termines de trabajar en tu proyecto o necesites volver al ambiente global, puedes desactivar el ambiente virtual con:
```bash
deactivate
```
Este comando te retorna al ambiente global de tu sistema, desactivando el ambiente virtual.






## Uso de `pipenv` para la gestión de ambientes virtuales

`pipenv` es una herramienta que automatiza la gestión de ambientes virtuales y dependencias en Python, facilitando la configuración de proyectos reproducibles y aislados. A continuación, te explicamos cómo usar `pipenv` en tus proyectos.

### 1. Instalar `pipenv`
Primero, necesitas instalar `pipenv` si aún no está instalado en tu sistema:
```bash
pip install pipenv
```

### 2. Preparar el entorno
Si ya tienes una carpeta de ambiente virtual (`env`) y deseas empezar de nuevo, es recomendable borrar esta carpeta antes de continuar.

### 3. Crear el ambiente virtual y añadir dependencias
Para crear un nuevo ambiente virtual e instalar una dependencia, como `requests`, puedes hacerlo con un solo comando:
```bash
pipenv install requests
```
Este comando crea un nuevo ambiente virtual por defecto fuera de tu carpeta de proyecto, para mantener organizado tu espacio de trabajo.

### 4. Localizar el ambiente virtual
Para saber dónde `pipenv` ha creado el ambiente virtual en tu disco duro, utiliza:
```bash
pipenv --venv
```
Este comando te mostrará la ruta completa del ambiente virtual.

### 5. Ignorar archivos innecesarios
Es prudente configurar un archivo `.gitignore` para evitar que se suban al repositorio archivos no necesarios. Normalmente, solo necesitas compartir el código fuente y los archivos `Pipfile` y `Pipfile.lock`, que contienen información sobre tus dependencias.

### 6. Activar el ambiente virtual
Para activar el ambiente virtual y comenzar a trabajar dentro de él, utiliza:
```bash
pipenv shell
```
Verás el nombre del ambiente virtual en tu prompt, indicando que cualquier comando de Python o `pip` que ejecutes operará dentro del ambiente.

### 7. Salir del ambiente virtual
Para salir del ambiente virtual y volver al entorno global de tu sistema, simplemente escribe:
```bash
exit
```
y presiona `Enter`. Esto desactivará el ambiente virtual.




## Uso de Pipfile y Pipfile.lock en proyectos con `pipenv`

Cuando inicializas un proyecto con `pipenv`, se crean dos archivos importantes: `Pipfile` y `Pipfile.lock`. Estos archivos son esenciales para gestionar las dependencias de tu proyecto de manera eficiente y reproducible.

### Pipfile

El `Pipfile` contiene la siguiente información:

- **[[source]]**: Indica la fuente de los paquetes, usualmente PyPI.
  ```toml
  url = "https://pypi.org/simple"
  ```

- **[packages]**: Lista los paquetes requeridos para que tu proyecto funcione.
  ```toml
  requests = "*"
  ```

- **[dev-packages]**: Incluye paquetes que son útiles para el desarrollo pero no necesarios para la ejecución de la aplicación.

- **[requires]**: Especifica la versión de Python necesaria para el proyecto.
  ```toml
  python_version = "3.8"
  ```

### Pipfile.lock

- El `Pipfile.lock` asegura que las dependencias se instalen exactamente de la misma forma cada vez, registrando las versiones exactas de cada paquete instalado. Esto ayuda a evitar inconsistencias entre entornos de desarrollo y producción.

### Flujo de trabajo recomendado con pipenv

1. **Visualización de la carpeta del ambiente virtual**:
   Para ver la ubicación del ambiente virtual creado por `pipenv`, puedes usar:
   ```bash
   pipenv --venv
   ```

2. **Eliminar la carpeta del ambiente virtual** (si es necesario):
   Si necesitas eliminar el ambiente virtual, primero localiza la carpeta con el comando anterior y luego usa:
   ```bash
   rm -rf <ruta_del_ambiente_virtual>
   ```
   **Advertencia**: Ten mucho cuidado al usar `rm -rf`, ya que podrías borrar archivos importantes si no se especifica correctamente la ruta.

3. **Instalación de dependencias**:
   Para instalar las dependencias listadas en tu `Pipfile`, simplemente ejecuta:
   ```bash
   pipenv install
   ```
   Esto creará un nuevo ambiente virtual e instalará todas las dependencias necesarias.

4. **Uso de Pipfile.lock para instalaciones consistentes**:
   Para asegurarte de que se instalen exactamente las versiones registradas en `Pipfile.lock`, ignora el `Pipfile` al instalar dependencias:
   ```bash
   pipenv install --ignore-pipfile
   ```









## Gestión de Dependencias con `pipenv`

`pipenv` proporciona varias herramientas útiles para la gestión de las dependencias de un proyecto Python, permitiéndote visualizar, actualizar y mantener el control sobre los paquetes que tu aplicación necesita.

### Visualizar dependencias

El comando `pipenv graph` muestra un árbol de todas las dependencias instaladas en tu proyecto, incluyendo las dependencias de cada paquete. Por ejemplo:

```
requests==2.28.2
   - urllib3 [required: >=1.21.1, <1.27, installed: 1.26.14]
```

Este output te ayuda a entender no solo qué paquetes están instalados, sino también las versiones específicas y las dependencias entre ellos.

### Desinstalar un paquete

Cuando desinstalas un paquete con:

```bash
pipenv uninstall requests
```

`pipenv` elimina el paquete y actualiza automáticamente el `Pipfile` y el `Pipfile.lock` para reflejar este cambio.

### Revisar dependencias residuales

A veces, después de desinstalar un paquete principal como `requests`, sus dependencias pueden no eliminarse automáticamente. Para verificar esto, puedes volver a ejecutar:

```bash
pipenv graph
```

Si aún ves dependencias que deberían haberse eliminado, puedes considerar desinstalarlas manualmente si no son utilizadas por otros paquetes.

### Actualizar paquetes

Para comprobar si hay actualizaciones disponibles para las dependencias instaladas:

```bash
pipenv update --outdated
```

Este comando muestra qué paquetes tienen versiones más recientes disponibles.

### Especificar versiones en `Pipfile`

Puedes especificar versiones específicas en tu `Pipfile` bajo la sección `[packages]` para asegurarte de que `pipenv` use la versión deseada:

```toml
requests = "==2.*"
```

Luego, al ejecutar:

```bash
pipenv update --outdated
```

`pipenv` ajustará las dependencias para alinearse con las versiones especificadas sin mostrar advertencias de desactualización.

### Actualizar todas las dependencias

Para actualizar todas las dependencias de tu proyecto a las versiones más recientes permitidas por las restricciones definidas en tu `Pipfile`:

```bash
pipenv update
```

### Actualizar un paquete específico

Si solo quieres actualizar un paquete específico, puedes hacerlo con:

```bash
pipenv update requests
```

Este comando actualizará `requests` y sus dependencias a las últimas versiones compatibles.









## Documentando paquetes en Python

### Crear un paquete y documentarlo

1. **Estructura del proyecto**: Comienza creando una estructura básica para tu paquete:

    ```
    calculadora
    ├── calculadora
    │   ├── __init__.py
    │   └── operaciones.py
    ├── tests
    ├── mocks
    ├── app.py
    ```

   - **mocks**: Carpeta para datos de prueba.
   - **tests**: Carpeta para pruebas unitarias del paquete.

2. **Creación de módulos**: Dentro de la subcarpeta `calculadora`, crea los siguientes archivos:
   - `__init__.py`: Hace que Python trate los directorios como conteniendo paquetes.
   - `operaciones.py`: Contiene las funciones del paquete.

3. **Archivo `app.py`**: En la raíz del proyecto, crea un archivo `app.py` para demostrar cómo usar el paquete:
    ```python
    from calculadora import operaciones

    # Crear una instancia de la calculadora
    calc = operaciones.Calculadora()
    resultado = calc.sumar(5, 3)
    print(resultado)  # Esto debería imprimir 8 en la consola
    ```

4. **Documentación del paquete**: En `__init__.py`, documenta el propósito del paquete:
    ```python
    """
    Este paquete proporciona operaciones matemáticas básicas como sumar, restar, etc.
    """
    ```

5. **Documentación del módulo `operaciones.py`**:
    ```python
    """
    Este módulo incluye una clase para realizar operaciones matemáticas básicas.
    """

    class Calculadora:
        """
        Clase para realizar operaciones matemáticas.
        """

        def sumar(self, a, b):
            """
            Suma dos números y devuelve el resultado.

            Parameters:
            a (int): Primer número entero.
            b (int): Segundo número entero.

            Returns:
            int: La suma de `a` y `b`.
            """
            return a + b

        def restar(self, a, b):
            """
            Resta el segundo número del primero y devuelve el resultado.

            Parameters:
            a (int): Primer número entero.
            b (int): Segundo número entero.

            Returns:
            int: La diferencia entre `a` y `b`.
            """
            return a - b
    ```









## Publicar paquetes en PyPI

### Registro y preparación
1. **Registrarse en PyPI**: Crea una cuenta en [PyPI](https://pypi.org/) si aún no tienes una.

2. **Instalación de herramientas necesarias**: Instala globalmente los paquetes necesarios para empaquetar y subir tu proyecto:
   ```bash
   pip install setuptools wheel twine
   ```

3. **Elección de licencia**: Visita [Choose a License](https://choosealicense.com/) para seleccionar una licencia adecuada para tu proyecto. Crea un archivo `LICENSE` y pega el texto de la licencia elegida.

4. **Crear README.md**: Este archivo debe contener una descripción de tu proyecto, cómo instalarlo, cómo usarlo y cualquier otra información relevante.

### Configuración del paquete
5. **Archivo setup.py**: Crea un archivo `setup.py` en la raíz de tu proyecto con el siguiente contenido:
   ```python
   import setuptools
   from pathlib import Path

   # Lee la descripción larga desde el archivo README.md
   long_desc = Path("README.md").read_text()

   setuptools.setup(
       name = "calculadora",
       version = "0.0.1",
       long_description = long_desc,
       long_description_content_type="text/markdown",
       packages = setuptools.find_packages(exclude=["tests", "mocks"])
   )
   ```

### Empaquetar y subir
6. **Preparar distribuciones**: Limpia tu terminal y luego crea las distribuciones del paquete:
   ```bash
   clear
   python setup.py sdist bdist_wheel
   ```
   - `sdist` significa distribución de código fuente.
   - `bdist_wheel` crea una distribución binaria.

   Esto generará carpetas `build`, `dist` y `calculadora.egg-info` en tu proyecto.

7. **Subir el paquete a PyPI**:
   ```bash
   twine upload dist/*
   ```
   Sigue las instrucciones para ingresar tu usuario y contraseña de PyPI.

### Uso del paquete
8. **Instalación y prueba**:
   - En un nuevo proyecto, crea una carpeta y un archivo `app.py`.
   - Instala tu paquete:
     ```bash
     pipenv install calculadora==0.0.1
     ```
   - En `app.py`, importa y usa tu módulo:
     ```python
     from calculadora.operaciones import Calculadora

     calc = Calculadora()
     print(calc.sumar(5, 3))  # Debería imprimir 8
     ```

   - Actualiza el ambiente virtual y asegúrate de que estás utilizando el ambiente correcto.

Esta guía te ayudará a publicar tu propio paquete en PyPI, facilitando su distribución y reutilización en la comunidad de Python.
