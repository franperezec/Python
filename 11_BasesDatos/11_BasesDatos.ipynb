# Bases de Datos SQL en Python

## Trabajando con Bases de Datos SQL en Python usando SQLite

SQLite es un motor de base de datos ligero y autónomo que no requiere un servidor para funcionar. Viene integrado en la biblioteca estándar de Python, lo que lo hace extremadamente conveniente para el desarrollo de aplicaciones pequeñas y medianas, pruebas, y prototipos. En esta sección, aprenderemos a trabajar con SQLite en Python, desde la creación de una base de datos hasta la manipulación de tablas y datos.

---

### Crear una Conexión a una Base de Datos SQLite

Lo primero que necesitas para trabajar con SQLite en Python es establecer una conexión a una base de datos. Si la base de datos especificada no existe, SQLite la creará automáticamente.

```python
import sqlite3  # Importamos el módulo sqlite3 para trabajar con bases de datos SQLite

# Crear una Conexión a la Base de Datos
# -------------------------------------
# Establecemos una conexión a la base de datos "ventas.db".
# Si la base de datos no existe, SQLite la crea automáticamente en el directorio especificado.
conexion = sqlite3.connect('11_BasesDAtos/ventas.db')

# Es importante cerrar la conexión a la base de datos para asegurarse de que todos los cambios se guarden y liberar recursos.
conexion.close()
```

- **`sqlite3.connect()`**: Crea una conexión a la base de datos SQLite. Si el archivo de la base de datos especificado no existe, se creará automáticamente.
- **`conexion.close()`**: Cierra la conexión a la base de datos, asegurando que todos los cambios se guarden y liberando los recursos utilizados.

### Crear y Manipular Tablas en SQLite

Una vez que se ha establecido la conexión a la base de datos, el siguiente paso es crear tablas y manipular los datos en ellas. Para interactuar con la base de datos, se utiliza un cursor.

```python
# Crear una Conexión a la Base de Datos y un Cursor
conexion = sqlite3.connect('11_BasesDAtos/ventas.db')
cursor = conexion.cursor()  # El cursor permite ejecutar sentencias SQL en la base de datos

# Crear una Tabla
# ----------------
# Aquí creamos una tabla llamada "ventas" para almacenar información sobre las ventas.
cursor.execute('''
    CREATE TABLE IF NOT EXISTS ventas (
        id INTEGER PRIMARY KEY,
        producto TEXT NOT NULL,
        cantidad INTEGER NOT NULL,
        precio REAL NOT NULL,
        fecha TEXT NOT NULL
    )
''')

# Confirmar los cambios en la base de datos
conexion.commit()

# Cerrar la Conexión a la Base de Datos
conexion.close()
```

- **`cursor.execute()`**: Ejecuta una sentencia SQL en la base de datos.
- **`CREATE TABLE IF NOT EXISTS`**: Crea una tabla solo si no existe ya en la base de datos. Esto es útil para evitar errores si intentas crear una tabla que ya existe.
- **`conexion.commit()`**: Confirma los cambios realizados en la base de datos durante la sesión.

### Insertar Datos en una Tabla

Después de crear una tabla, puedes empezar a insertar datos en ella. Aquí se muestra cómo agregar una fila de datos a la tabla "ventas".

```python
# Conectar a la base de datos y crear un cursor
conexion = sqlite3.connect('11_BasesDAtos/ventas.db')
cursor = conexion.cursor()

# Insertar Datos en la Tabla
cursor.execute('''
    INSERT INTO ventas (producto, cantidad, precio, fecha)
    VALUES ('Manzanas', 10, 0.5, '2024-08-21')
''')

# Confirmar los cambios en la base de datos
conexion.commit()

# Cerrar la conexión
conexion.close()
```

- **`INSERT INTO ventas`**: Inserta una nueva fila en la tabla "ventas" con los valores especificados.

### Consultar Datos desde una Tabla

Para recuperar datos almacenados en la base de datos, se utiliza la sentencia `SELECT`. Este comando permite seleccionar filas de una tabla para su visualización o procesamiento.

```python
# Conectar a la base de datos y crear un cursor
conexion = sqlite3.connect('11_BasesDAtos/ventas.db')
cursor = conexion.cursor()

# Consultar los Datos de la Tabla
cursor.execute('SELECT * FROM ventas')  # Selecciona todas las filas de la tabla "ventas"
resultados = cursor.fetchall()  # Obtiene todas las filas como una lista de tuplas

# Imprimir los Resultados
for fila in resultados:
    print(fila)

# Cerrar la conexión
conexion.close()
```

- **`SELECT * FROM ventas`**: Selecciona todas las columnas de todas las filas en la tabla "ventas".
- **`cursor.fetchall()`**: Devuelve todas las filas seleccionadas por la consulta como una lista de tuplas, donde cada tupla representa una fila.

SQLite es una opción excelente para proyectos que requieren una base de datos ligera y fácil de manejar. No requiere configuración adicional y se integra perfectamente con Python, lo que lo hace ideal para el desarrollo rápido de prototipos y aplicaciones pequeñas. ## Creación de Tablas en una Base de Datos SQLite

Cuando trabajamos con bases de datos, una de las primeras tareas que realizamos es la creación de tablas. Las tablas son estructuras fundamentales donde se almacenan los datos en la base de datos. En Python, podemos utilizar el módulo `sqlite3` para interactuar con una base de datos SQLite, incluyendo la creación de tablas. A continuación, te guiaré a través del proceso de creación de una tabla en SQLite utilizando Python.

---

### Código para Crear una Tabla

```python
import sqlite3  # Importar el módulo de sqlite3 para trabajar con bases de datos SQLite

# Crear una conexión a la base de datos
# -------------------------------------
# Establecemos una conexión a la base de datos "ventas.db".
# Si el archivo "ventas.db" no existe, SQLite lo crea automáticamente.
conexion = sqlite3.connect('11_BasesDatos/ventas.db')

# Crear un cursor para ejecutar sentencias SQL
# --------------------------------------------
# El cursor es un objeto que nos permite interactuar con la base de datos,
# ejecutar sentencias SQL y obtener resultados.
cursor = conexion.cursor()

# Crear una Tabla en la Base de Datos
# -----------------------------------
# Aquí estamos creando una tabla llamada "productos" en la base de datos "ventas.db".
# Esta tabla tendrá las siguientes columnas:
# - id: un identificador único para cada producto (clave primaria).
# - nombre: el nombre del producto.
# - precio: el precio del producto.
cursor.execute('''
    CREATE TABLE IF NOT EXISTS productos (
        id INTEGER PRIMARY KEY,
        nombre VARCHAR(100),
        precio FLOAT
    )
''')

# Guardar los Cambios en la Base de Datos
# ---------------------------------------
# La llamada a `commit()` es crucial porque confirma los cambios realizados en la base de datos.
# Esto incluye la creación de tablas o la inserción de datos. Sin `commit()`, los cambios no se aplicarán.
conexion.commit()

# Cerrar la Conexión a la Base de Datos
# -------------------------------------
# Siempre es importante cerrar la conexión a la base de datos una vez que hemos terminado de trabajar con ella.
# Esto libera los recursos y asegura que todos los cambios se guarden correctamente.
conexion.close()
```

---

## Explicación Detallada

1. **Importar el Módulo `sqlite3`**:
   - El módulo `sqlite3` permite que Python se conecte a una base de datos SQLite y ejecute sentencias SQL.

2. **Crear una Conexión a la Base de Datos**:
   - Utilizamos `sqlite3.connect()` para establecer una conexión a la base de datos. Si la base de datos especificada no existe, SQLite la crea automáticamente.

3. **Crear un Cursor**:
   - El cursor es un objeto utilizado para enviar comandos SQL a la base de datos. Es necesario para ejecutar cualquier sentencia SQL.

4. **Crear una Tabla**:
   - Utilizamos la sentencia `CREATE TABLE IF NOT EXISTS` para crear una tabla llamada "productos". La cláusula `IF NOT EXISTS` asegura que la tabla solo se creará si no existe ya, evitando posibles errores.

5. **Guardar los Cambios con `commit()`**:
   - La función `commit()` se utiliza para confirmar los cambios realizados en la base de datos, como la creación de tablas o la inserción de datos. Si no se llama a `commit()`, los cambios no se guardarán.

6. **Cerrar la Conexión**:
   - Cerrar la conexión es una buena práctica para liberar recursos y asegurar que todos los cambios se hayan guardado correctamente en la base de datos.

## Uso de `with` en la Gestión de Conexiones a Bases de Datos

El contexto `with` en Python es una estructura que simplifica la gestión de recursos como archivos, conexiones a bases de datos, y otros recursos que requieren ser abiertos y luego cerrados. Cuando se utiliza `with` para manejar conexiones a bases de datos, Python se encarga automáticamente de realizar el `commit` y cerrar la conexión cuando se sale del bloque `with`, lo que hace que el código sea más limpio y menos propenso a errores.

---

## Ejemplo de Uso de `with` para Conexiones a Bases de Datos SQLite

En este ejemplo, usaremos el contexto `with` para manejar la conexión a una base de datos SQLite. La ventaja de usar `with` es que no necesitamos preocuparnos por cerrar la conexión manualmente; Python lo hará por nosotros al finalizar el bloque.

```python
import sqlite3  # Importamos el módulo sqlite3 para trabajar con bases de datos SQLite

# Crear una conexión a la base de datos utilizando `with`
# -------------------------------------------------------
# La conexión se crea y se asigna a la variable `conexion`.
# Cuando el bloque `with` se complete, la conexión se cerrará automáticamente y cualquier cambio se guardará.
with sqlite3.connect('11_BasesDatos/ventas.db') as conexion:

    # Crear un cursor para ejecutar sentencias SQL
    cursor = conexion.cursor()

    # Crear una tabla llamada "productos" si no existe
    # ------------------------------------------------
    # Utilizamos la sentencia SQL `CREATE TABLE IF NOT EXISTS` para crear la tabla "productos".
    # La tabla tendrá tres columnas:
    # - id: un identificador único para cada producto (clave primaria).
    # - nombre: el nombre del producto.
    # - precio: el precio del producto.
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS productos (
            id INTEGER PRIMARY KEY,
            nombre VARCHAR(100),
            precio FLOAT
        )
    ''')  # Ejecutamos la sentencia SQL para crear la tabla.
```

## Explicación del Código

1. **Importar el Módulo `sqlite3`**:
   - Como en los ejemplos anteriores, el módulo `sqlite3` es necesario para interactuar con bases de datos SQLite en Python.

2. **Uso de `with` para la Conexión a la Base de Datos**:
   - Utilizar `with` para manejar la conexión a la base de datos es una práctica recomendada, ya que asegura que la conexión se cierre automáticamente una vez que se termina el bloque de código dentro de `with`. Esto incluye la ejecución de un `commit` si se han realizado cambios en la base de datos, lo que asegura que los cambios se guarden.

3. **Creación de un Cursor**:
   - El cursor se utiliza para ejecutar sentencias SQL en la base de datos. Dentro del bloque `with`, puedes ejecutar cualquier número de sentencias SQL utilizando el cursor.

4. **Creación de una Tabla**:
   - En este ejemplo, se crea una tabla llamada "productos" con tres columnas: `id`, `nombre`, y `precio`. La columna `id` es la clave primaria, lo que significa que cada registro tendrá un identificador único.

## Ventajas de Usar `with` para Conexiones a Bases de Datos

- **Manejo Automático de Recursos**: El uso de `with` asegura que los recursos se manejen correctamente sin necesidad de escribir código adicional para cerrar la conexión o confirmar transacciones.
- **Código Más Limpio y Claro**: Al no tener que preocuparnos por cerrar la conexión manualmente, el código se vuelve más legible y menos propenso a errores.
- **Seguridad en la Gestión de Errores**: Si ocurre una excepción dentro del bloque `with`, Python aún cerrará la conexión correctamente, evitando posibles bloqueos o pérdidas de datos.

El uso de `with` es una técnica poderosa que te permite escribir código más seguro y eficiente cuando trabajas con bases de datos en Python.

## Inserción de Datos en una Base de Datos SQLite usando `INSERT`

La inserción de datos en una base de datos es una de las operaciones más comunes al trabajar con bases de datos relacionales. En Python, utilizando `sqlite3`, puedes insertar datos de manera segura y eficiente en tablas de una base de datos SQLite. A continuación, exploraremos cómo realizar una inserción de datos en una tabla usando la sentencia `INSERT INTO`.

---

### Código para Insertar Datos en una Tabla

El siguiente código demuestra cómo insertar un nuevo registro en la tabla "productos" de la base de datos "ventas.db":

```python
import sqlite3  # Importamos el módulo sqlite3 para trabajar con bases de datos SQLite

# Crear una conexión a la base de datos usando el contexto `with`
# ---------------------------------------------------------------
# Utilizamos `with` para abrir una conexión a la base de datos "ventas.db" ubicada en el directorio "11_BasesDatos".
# Si la base de datos no existe, SQLite la creará automáticamente.
with sqlite3.connect('11_BasesDatos/ventas.db') as conexion:
    # Crear un cursor para ejecutar sentencias SQL
    cursor = conexion.cursor()

    # Insertar datos en la tabla "productos"
    # --------------------------------------
    # Usamos la sentencia SQL `INSERT INTO` para agregar un nuevo registro en la tabla "productos".
    # Los valores a insertar son: id = 1, nombre = "Teclado", precio = 20.0.
    # La sintaxis `?` es un marcador de posición que será reemplazado por los valores proporcionados en la tupla.
    # Esto nos ayuda a prevenir inyecciones SQL, ya que los valores son pasados como un segundo argumento separado.
    cursor.execute(
        "INSERT INTO productos VALUES(?, ?, ?)",
        (1, "Teclado", 20.0)
    )
```

## Explicación del Código

1. **Importación del Módulo `sqlite3`**:
   - Al igual que en otros ejemplos, utilizamos el módulo `sqlite3` para interactuar con bases de datos SQLite desde Python.

2. **Uso del Contexto `with`**:
   - El uso de `with` para manejar la conexión a la base de datos asegura que la conexión se cerrará automáticamente al final del bloque `with`, y cualquier cambio realizado, como la inserción de un nuevo registro, se guardará automáticamente.

3. **Creación de un Cursor**:
   - El cursor se utiliza para ejecutar la sentencia SQL `INSERT INTO`, que añade un nuevo registro a la tabla "productos".

4. **Sentencia `INSERT INTO`**:
   - **`INSERT INTO productos VALUES(?, ?, ?)`**: Esta sentencia SQL se utiliza para insertar un nuevo registro en la tabla "productos".
   - **Marcadores de Posición (`?`)**: Los `?` son marcadores de posición que permiten la inserción de valores de manera segura. Estos marcadores serán reemplazados por los valores que se pasan como segundo argumento en la tupla `(1, "Teclado", 20.0)`.
   - **Prevención de Inyecciones SQL**: Al usar marcadores de posición y pasar los valores como un segundo argumento separado, se evita que usuarios malintencionados puedan inyectar código SQL malicioso.

## Importancia de la Seguridad en la Inserción de Datos

El uso de marcadores de posición (`?`) es una práctica recomendada para prevenir inyecciones SQL, que son uno de los ataques más comunes en aplicaciones web que interactúan con bases de datos. Las inyecciones SQL ocurren cuando un atacante inserta código SQL malicioso en una entrada que es directamente ejecutada por la base de datos. Al utilizar parámetros preparados, como en el ejemplo anterior, puedes asegurar que los valores insertados sean tratados como datos y no como parte de la sentencia SQL.

## Conclusión

Insertar datos en una base de datos es una operación fundamental que debe realizarse de manera segura para proteger la integridad de la base de datos. En este ejemplo, has aprendido cómo usar la sentencia `INSERT INTO` con marcadores de posición para agregar un registro a una tabla en SQLite, asegurando que los datos se manejen de manera segura y eficiente.

## Inserción Múltiple de Registros en una Base de Datos SQLite usando `executemany`

Cuando trabajas con bases de datos, a menudo necesitas insertar múltiples registros de manera eficiente. En lugar de ejecutar una sentencia `INSERT INTO` para cada registro individualmente, puedes utilizar la función `executemany()` de `sqlite3`. Esta función te permite realizar inserciones masivas en una tabla de una sola vez, lo que mejora el rendimiento y reduce el tiempo de ejecución.

---

### Código para Realizar una Inserción Múltiple en una Tabla

El siguiente código demuestra cómo insertar varios registros en la tabla "productos" de la base de datos "ventas.db":

```python
import sqlite3  # Importamos el módulo sqlite3 para trabajar con bases de datos SQLite

# Crear una conexión a la base de datos usando el contexto `with`
# ---------------------------------------------------------------
# Utilizamos `with` para abrir una conexión a la base de datos "ventas.db" ubicada en el directorio "11_BasesDatos".
# Si el archivo de la base de datos "ventas.db" no existe, SQLite lo creará automáticamente.
with sqlite3.connect('11_BasesDatos/ventas.db') as conexion:

    # Crear un cursor para ejecutar sentencias SQL
    cursor = conexion.cursor()

    # Lista de productos a insertar en la tabla
    # -----------------------------------------
    # Aquí creamos una lista de tuplas, donde cada tupla representa un producto que queremos insertar en la tabla "productos".
    # Cada tupla contiene los valores: id, nombre y precio.
    productos = [
        (2, "Ratón", 10.0),
        (3, "Monitor", 200.0),
        (4, "Impresora", 150.0),
    ]

    # Insertar múltiples registros en la tabla "productos"
    # ----------------------------------------------------
    # Utilizamos la función `executemany()` para insertar varios registros a la vez.
    # La sentencia SQL `INSERT INTO productos VALUES(?, ?, ?)` se aplica a cada tupla en la lista `productos`.
    # Los `?` son marcadores de posición que serán reemplazados por los valores de cada tupla en la lista.
    cursor.executemany(
        "INSERT INTO productos VALUES(?, ?, ?)",
        productos
    )
```

## Explicación del Código

1. **Importación del Módulo `sqlite3`**:
   - Como en ejemplos anteriores, el módulo `sqlite3` se utiliza para interactuar con bases de datos SQLite en Python.

2. **Uso del Contexto `with` para la Conexión a la Base de Datos**:
   - `with` se utiliza para manejar la conexión a la base de datos. Esto asegura que la conexión se cierre automáticamente al finalizar el bloque `with`, y que cualquier cambio realizado se guarde correctamente.

3. **Creación de un Cursor**:
   - El cursor es necesario para ejecutar sentencias SQL en la base de datos. En este caso, se utiliza para insertar varios registros en la tabla "productos".

4. **Lista de Productos**:
   - Creamos una lista llamada `productos`, donde cada elemento es una tupla que contiene los valores `id`, `nombre`, y `precio` de un producto. Esta lista se pasará a la función `executemany()` para realizar la inserción múltiple.

5. **Uso de `executemany()` para Inserciones Múltiples**:
   - **`executemany()`**: Esta función permite ejecutar la misma sentencia SQL varias veces, utilizando diferentes conjuntos de datos. En este caso, la sentencia `INSERT INTO productos VALUES(?, ?, ?)` se ejecuta para cada tupla en la lista `productos`.
   - **Marcadores de Posición (`?`)**: Los marcadores de posición `?` en la sentencia SQL se reemplazan por los valores correspondientes en cada tupla de la lista `productos`. Esto asegura que los datos se inserten correctamente y de manera segura.

## Ventajas de Usar `executemany()`

- **Eficiencia**: Insertar múltiples registros a la vez con `executemany()` es mucho más eficiente que ejecutar una sentencia `INSERT INTO` por separado para cada registro. Esto reduce el número de operaciones realizadas en la base de datos, mejorando el rendimiento.
- **Código más Limpio**: Al agrupar las inserciones en una sola operación, el código es más limpio y fácil de mantener.
- **Seguridad**: Al igual que en las inserciones individuales, el uso de marcadores de posición evita las inyecciones SQL, asegurando que los valores insertados sean tratados como datos.

## Conclusión

La inserción múltiple de registros es una técnica esencial para mejorar la eficiencia al trabajar con bases de datos. El uso de `executemany()` en `sqlite3` permite insertar varios registros de manera rápida y segura, asegurando que tu aplicación funcione de manera óptima incluso cuando maneje grandes volúmenes de datos.

## Selección de un Solo Registro en una Base de Datos SQLite usando `SELECT`

En muchas ocasiones, al trabajar con bases de datos, necesitamos recuperar un solo registro de una tabla. Esto puede hacerse utilizando la sentencia `SELECT` junto con el método `fetchone()` de un cursor. Este enfoque es útil cuando sabes que la consulta SQL devolverá un único registro o si solo estás interesado en el primer resultado de un conjunto de resultados.

---

### Código para Seleccionar un Solo Registro

El siguiente código muestra cómo recuperar un solo registro de la tabla "productos" en la base de datos "ventas.db":

```python
import sqlite3  # Importamos el módulo sqlite3 para trabajar con bases de datos SQLite

# Crear una conexión a la base de datos usando el contexto `with`
# ---------------------------------------------------------------
# Utilizamos `with` para abrir una conexión a la base de datos "ventas.db" ubicada en el directorio "11_BasesDatos".
# Si la base de datos no existe, SQLite la creará automáticamente.
with sqlite3.connect('11_BasesDatos/ventas.db') as conexion:

    # Crear un cursor para ejecutar sentencias SQL
    cursor = conexion.cursor()

    # Seleccionar todos los registros de la tabla "productos"
    # -------------------------------------------------------
    # Ejecutamos una sentencia SQL `SELECT * FROM productos` que selecciona todas las columnas de todos los registros en la tabla "productos".
    cursor.execute('SELECT * FROM productos')

    # Recuperar e imprimir el primer registro del resultado
    # -----------------------------------------------------
    # Utilizamos el método `fetchone()` del cursor para obtener el primer registro del conjunto de resultados.
    # `fetchone()` devuelve una tupla que contiene los valores del registro.
    # Si no hay más registros disponibles, devuelve `None`.
    print(cursor.fetchone())  # Imprime el primer resultado de la consulta
```

## Explicación del Código

1. **Importación del Módulo `sqlite3`**:
   - El módulo `sqlite3` permite interactuar con bases de datos SQLite desde Python. Es parte de la biblioteca estándar de Python, por lo que no es necesario instalar nada adicional.

2. **Uso del Contexto `with` para la Conexión a la Base de Datos**:
   - Al utilizar `with`, la conexión a la base de datos se gestiona automáticamente. Se asegura que la conexión se cierra adecuadamente cuando se sale del bloque `with`, incluso si ocurre un error.

3. **Creación de un Cursor**:
   - El cursor es un objeto necesario para ejecutar sentencias SQL y recuperar resultados de la base de datos.

4. **Sentencia `SELECT * FROM productos`**:
   - La sentencia SQL `SELECT * FROM productos` selecciona todas las columnas de todos los registros en la tabla "productos". El asterisco (`*`) indica que se seleccionarán todas las columnas.

5. **Uso de `fetchone()` para Recuperar un Solo Registro**:
   - El método `fetchone()` del cursor se utiliza para obtener el primer registro del conjunto de resultados de la consulta. Devuelve una tupla con los valores de ese registro. Si no hay más registros disponibles, `fetchone()` devolverá `None`.
   - Este método es útil cuando solo necesitas un registro, como por ejemplo, para verificar si existe un registro que cumpla con ciertas condiciones.

## Consideraciones Adicionales

- **Uso Común**: `fetchone()` es comúnmente usado cuando esperas que la consulta devuelva solo un registro o si te interesa solo el primer registro del resultado.
- **Optimización**: Al recuperar solo el primer registro, puedes ahorrar memoria y mejorar el rendimiento en consultas donde no necesitas todos los resultados.

## Conclusión

El método `fetchone()` es una herramienta simple pero poderosa para recuperar un solo registro de una base de datos. Es especialmente útil en situaciones donde solo necesitas verificar la existencia de un registro o cuando te interesa únicamente el primer resultado de una consulta. Utilizarlo dentro de un contexto `with` asegura que la conexión a la base de datos se maneje de manera eficiente y segura.

## Selección de Todos los Registros en una Base de Datos SQLite usando `SELECT`

Cuando trabajas con bases de datos, es común que necesites recuperar todos los registros de una tabla para analizarlos, mostrarlos en una interfaz o procesarlos de alguna manera. En SQLite, puedes lograr esto usando la sentencia `SELECT` junto con el método `fetchall()` de un cursor. Este método te permite obtener todos los resultados de una consulta SQL en una sola operación.

### Código para Seleccionar Todos los Registros

El siguiente código muestra cómo recuperar todos los registros de la tabla "productos" en la base de datos "ventas.db":

```python
import sqlite3  # Importamos el módulo sqlite3 para trabajar con bases de datos SQLite

# Crear una conexión a la base de datos usando el contexto `with`
# ---------------------------------------------------------------
# Utilizamos `with` para abrir una conexión a la base de datos "ventas.db" ubicada en el directorio "11_BasesDatos".
# Si la base de datos "ventas.db" no existe, SQLite la creará automáticamente.
with sqlite3.connect('11_BasesDatos/ventas.db') as conexion:

    # Crear un cursor para ejecutar sentencias SQL
    cursor = conexion.cursor()

    # Seleccionar todos los registros de la tabla "productos"
    # -------------------------------------------------------
    # Ejecutamos una sentencia SQL `SELECT * FROM productos` que selecciona todas las columnas de todos los registros en la tabla "productos".
    cursor.execute('SELECT * FROM productos')

    # Recuperar e imprimir todos los registros del resultado
    # ------------------------------------------------------
    # Utilizamos el método `fetchall()` del cursor para obtener todos los registros del conjunto de resultados de la consulta.
    # `fetchall()` devuelve una lista de tuplas, donde cada tupla representa un registro de la tabla.
    print(cursor.fetchall())  # Imprime todos los resultados de la consulta
```

## Explicación del Código

1. **Importación del Módulo `sqlite3`**:
   - El módulo `sqlite3` es parte de la biblioteca estándar de Python y se utiliza para interactuar con bases de datos SQLite. Este módulo permite realizar operaciones como consultas SQL, inserciones, actualizaciones y más.

2. **Uso del Contexto `with` para la Conexión a la Base de Datos**:
   - Utilizar `with` para abrir la conexión a la base de datos garantiza que la conexión se cerrará automáticamente cuando se termine el bloque `with`. Esto es esencial para liberar recursos y asegurar que todos los cambios se guarden adecuadamente.

3. **Creación de un Cursor**:
   - El cursor es un objeto que se utiliza para ejecutar sentencias SQL y recuperar resultados de la base de datos. En este caso, se utiliza para ejecutar la consulta `SELECT` y recuperar todos los registros.

4. **Sentencia SQL `SELECT * FROM productos`**:
   - La sentencia SQL `SELECT * FROM productos` selecciona todas las columnas de todos los registros en la tabla "productos". El asterisco (`*`) se utiliza como comodín para indicar que se deben seleccionar todas las columnas.

5. **Uso de `fetchall()` para Recuperar Todos los Registros**:
   - El método `fetchall()` del cursor se utiliza para recuperar todos los registros resultantes de la consulta SQL. Este método devuelve una lista de tuplas, donde cada tupla representa un registro de la tabla.
   - **Salida**: Al imprimir el resultado de `fetchall()`, verás una lista con todos los registros de la tabla "productos". Cada registro se muestra como una tupla con los valores de sus columnas.

## Consideraciones Adicionales

- **Memoria**: `fetchall()` recupera todos los registros en la memoria, lo que puede ser un problema si la tabla contiene una gran cantidad de datos. Si esperas que la consulta devuelva un gran número de registros, es recomendable utilizar `fetchone()` o `fetchmany()` para recuperar los datos en partes.

- **Uso Común**: `fetchall()` es útil cuando necesitas procesar o mostrar todos los registros de una tabla, como en informes, exportaciones de datos, o al mostrar todos los resultados en una interfaz de usuario.

## Conclusión

El método `fetchall()` es una herramienta poderosa para recuperar todos los registros de una tabla en una base de datos SQLite. Al combinarlo con una consulta `SELECT`, puedes obtener y manipular todos los datos almacenados en una tabla de manera eficiente. Sin embargo, es importante tener en cuenta el tamaño de los datos que se están recuperando para asegurar un rendimiento óptimo.
